````python
# If this import fails, install the project dependencies / select the agents kernel.
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv(override=True)
````

````python
# Confirm that the Groq key is available without printing the secret itself.
groq_api_key = os.getenv('GROQ_API_KEY')

if groq_api_key:
    print(f'Groq API key exists and begins {groq_api_key[:8]}')
else:
    print('GROQ_API_KEY is not set. Add it to the .env file in the agents project root.')
````

````python
# Create an OpenAI-compatible client that sends requests to Groq.
groq = OpenAI(
    api_key=groq_api_key,
    base_url='https://api.groq.com/openai/v1',
)

# This is the Groq model used throughout this notebook.
MODEL = 'openai/gpt-oss-20b'
````

````python
# Create a list of messages in the familiar OpenAI chat format.
messages = [{'role': 'user', 'content': 'Tell me a fun fact'}]
messages
````

````python

#question_prompt = (
#    "Please propose a hard, challenging question to assess someone's IQ. "
#    "Respond only with the question."
#)
#messages = [{'role': 'user', 'content': question_prompt}]
#response = groq.chat.completions.create(model=MODEL, messages=messages)
#question = response.choices[0].message.content  <= qroq does not have this
#print(question)
````

````python
response = groq.chat.completions.create(
    model=MODEL,
    messages=messages,
    max_completion_tokens=512,
    extra_body={
        "include_reasoning": False,
        "reasoning_effort": "low",
    },
)

choice = response.choices[0]
print("finish reason:", choice.finish_reason)
print("content:", repr(choice.message.content))
````

````python
response = groq.chat.completions.create(
    model=MODEL,
    messages=messages,
    max_completion_tokens=512,
    extra_body={
        "include_reasoning": False,
        "reasoning_effort": "low",
    },
)

choice = response.choices[0]
print("finish reason:", choice.finish_reason)
print("content:", repr(choice.message.content))
print("full message:", choice.message)
````
